```python
# Import Pandas for data analysis
import pandas as pd
# Import Matplotlib for creating charts
import matplotlib.pyplot as plt
# Import file
from google.colab import files
# Upload a file
uploaded = files.upload()
```

```python
# Read the CSV file into a DataFrame
df = pd.read_csv('insurance_cleaned.csv')
```

```python
# Count the number of customers for each policy type
df['Policy_Type'].value_counts()
```

```python
# Create a bar chart showing the number of customers by policy type
df['Policy_Type'].value_counts().plot(kind='bar')

plt.title('Customers by Policy Type')
plt.xlabel('Policy Type')
plt.ylabel('Number of Customers')
plt.show()
```

```python
# Calculate the average premium for each policy typ
df.groupby('Policy_Type')['Premium'].mean()
```

```python
df.groupby('Policy_Type')['Premium'].mean().sort_values().plot(kind='bar')

plt.title('Average Premium by Policy Type')
plt.xlabel('Policy Type')
plt.ylabel('Average Premium')
plt.show()
```

```python
# Calculate the total number of claims for each policy type
df.groupby('Policy_Type')['Claim_Count'].sum()
```

```python
df.groupby('Policy_Type')['Claim_Count'].sum().sort_values().plot(kind='bar')

plt.title('Total Claims by Policy Type')
plt.xlabel('Policy Type')
plt.ylabel('Total Claims')
plt.show()
```
